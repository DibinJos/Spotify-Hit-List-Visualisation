```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data_df = pd.read_csv("./SpotifyTopSongsByCountry - May 2020.csv")
```

```python
uniq = data_df['Country'].unique()
```

```python
data_df_filter = data_df[data_df['Country'].isin(uniq) & (data_df['Rank'] == 1) & (data_df['Country'] != 'Global')] 
```

```python
data_df_filter['Country'].unique()
```

```python
from geopy.geocoders import Nominatim
import folium
```

```python
geolocator = Nominatim(user_agent='music_geocoder')
```

```python
def geocoder_address(row):
    print(row['Country'])
    location = geolocator.geocode(row['Country'])
    print("location is",location)
    if location:
        print("Success")
        return location.latitude, location.longitude
    else:
        print("Failure")
        return None, None
```

```python
data_df_filter['Latitude'], data_df_filter['Longitude'] = zip(*data_df_filter.apply(geocoder_address,axis=1))
```

```python
# Calculate the map center based on the points
map_center = [data_df_filter['Latitude'].mean(), data_df_filter['Longitude'].mean()]
```

```python
map_center
```

```python
mymap = folium.Map(location = map_center, zoom_start = 4)
```

```python
for index, row in data_df_filter.iterrows():
    
    if(row['Latitude'] and row['Longitude']):
        
        popup_text = f"""
        <b><span style='color:red;'>{row['Title']}</span></b>  
        <br>by <span style='color:blue;'>{row['Artists']}</span>  
        <br>in album <b>{row['Album']}</b>
        """
            
        folium.Marker(
            [row['Latitude'], row['Longitude']] , 
            icon=folium.Icon(color='blue', icon='music', prefix='fa'), 
            popup=folium.Popup(popup_text, max_width=300)
        ).add_to(mymap)
        
```

```python
mymap.save('music_geocode.html')
```
